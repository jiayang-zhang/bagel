```python
import bagel as bg
from biotite.structure.io.pdbx import CIFFile, set_structure
from biotite.structure.io import save_structure
from pathlib import Path
import os

sequence = 'MSKGEELFTGVVPILVELDGDVNGHKFSVSGEGEGDATYGKLTLKFICTTGKLPVPWPTLVTTFSYGVQCFSRYPDHMKQHDFFKSAMPEGYVQERTIFFKDDGNYKTRAEVKFEGDTLVNRIELKGIDFKEDGNILGHKLEYNYNSHNVYIMADKQKNGIKVNFKIRHNIEDGSVQLADHYQQNTPIGDGPVLLPDNHYLSTQSALSKDPNEKRDHMVLLEFVTAAGITHGMDELYK'
ID = "P42212"      
seq = sequence     


residues = [
    bg.Residue(name=aa, chain_ID='b', index=i, mutable=False)
    for i, aa in enumerate(seq)
]
chain = bg.Chain(residues=residues)

config = {
    "output_pdb": True,
    "output_cif": True,
}
esm = bg.oracles.ESMFold(use_modal=False, config=config)
result = esm.fold([chain])

# 如果你想自己控制输出路径，再手动写 pdb/cif：
out_dir = Path(os.getcwd())
out_dir.mkdir(parents=True, exist_ok=True)
out_stem = out_dir / f"seqID_{ID}"

structure_file = CIFFile()
set_structure(structure_file, result.structure)
structure_file.write(str(out_stem.with_suffix(".cif")))

# save_structure(str(out_stem.with_suffix(".pdb")), result.structure)

# result.save_attributes(out_stem)  # 写 .plddt / .pae

```
